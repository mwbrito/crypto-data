```python
import re
from pathlib import Path
from zipfile import ZipFile

diretorio_downloads = Path("downloads")
padrao_trades = re.compile(r"BTCUSDT-trades-\d{4}-(0[1-9]|1[0-2])\.zip$")
arquivos_trades = sorted(
    caminho for caminho in diretorio_downloads.glob("BTCUSDT-trades-*.zip")
    if padrao_trades.fullmatch(caminho.name)
)
if not arquivos_trades:
    raise FileNotFoundError(
        f"Nenhum arquivo BTCUSDT-trades-YYYY-MM.zip encontrado em "
        f"{diretorio_downloads.resolve()}"
    )

total_linhas_geral = 0
for arquivo_trades in arquivos_trades:
    with ZipFile(arquivo_trades) as zip_trades:
        arquivos_csv = [nome for nome in zip_trades.namelist() if nome.endswith(".csv")]
        if not arquivos_csv:
            raise ValueError(f"O ZIP não contém um arquivo CSV: {arquivo_trades}")

        total_linhas = 0
        ultimo_byte = b""
        with zip_trades.open(arquivos_csv[0]) as csv_trades:
            while bloco := csv_trades.read(8 * 1024 * 1024):
                total_linhas += bloco.count(b"\n")
                ultimo_byte = bloco[-1:]
        if ultimo_byte and ultimo_byte != b"\n":
            total_linhas += 1

    total_linhas_geral += total_linhas
    print(f"{arquivo_trades.name}: {total_linhas:,} linhas")

print(f"Total geral: {total_linhas_geral:,} linhas em {len(arquivos_trades)} arquivo(s)")
```

```python
import re
from contextlib import ExitStack
from pathlib import Path
from tempfile import TemporaryDirectory
from zipfile import ZipFile

import pandas as pd

try:
    import pyarrow as pa
    import pyarrow.parquet as pq
except ImportError as erro:
    raise ImportError("Instale o suporte a Parquet executando `%pip install pyarrow` em uma célula.") from erro

padrao_arquivo = re.compile(r"BTCUSDT-1s-(\d{4})-(\d{2})\.zip$")
diretorio_downloads = Path("downloads")
arquivos_mensais = sorted(
    caminho for caminho in diretorio_downloads.glob("BTCUSDT-1s-*.zip")
    if padrao_arquivo.fullmatch(caminho.name)
)
if not arquivos_mensais:
    raise FileNotFoundError(
        f"Nenhum ZIP BTCUSDT-1s-YYYY-MM encontrado em {diretorio_downloads.resolve()}"
    )

colunas_klines = [
    "open_time", "open", "high", "low", "close", "volume",
    "close_time", "quote_asset_volume", "number_of_trades",
    "taker_buy_base_asset_volume", "taker_buy_quote_asset_volume", "ignore",
]
tipos_klines = {
    "open_time": "int64", "open": "float64", "high": "float64",
    "low": "float64", "close": "float64", "volume": "float64",
    "close_time": "int64", "quote_asset_volume": "float64",
    "number_of_trades": "int64", "taker_buy_base_asset_volume": "float64",
    "taker_buy_quote_asset_volume": "float64", "ignore": "float64",
}
diretorio_parquet = Path("parquet_anual")
diretorio_parquet.mkdir(parents=True, exist_ok=True)
anos = sorted({padrao_arquivo.fullmatch(caminho.name).group(1) for caminho in arquivos_mensais})
# Converte em blocos para evitar carregar um ano inteiro de candles na memória.
with TemporaryDirectory(dir=diretorio_parquet) as pasta_temporaria:
    pasta_temporaria = Path(pasta_temporaria)
    arquivos_anuais_temp = {
        ano: pasta_temporaria / f"BTCUSDT-1s-{ano}.parquet" for ano in anos
    }
    with ExitStack() as pilha:
        gravadores = {}
        for arquivo_zip in arquivos_mensais:
            ano = padrao_arquivo.fullmatch(arquivo_zip.name).group(1)
            with ZipFile(arquivo_zip) as zip_mensal:
                membro_csv = next(nome for nome in zip_mensal.namelist() if nome.endswith(".csv"))
                with zip_mensal.open(membro_csv) as csv_mensal:
                    blocos = pd.read_csv(
                        csv_mensal,
                        header=None,
                        names=colunas_klines,
                        dtype=tipos_klines,
                        chunksize=250_000,
                    )
                    for bloco in blocos:
                        tabela = pa.Table.from_pandas(bloco, preserve_index=False)
                        if ano not in gravadores:
                            gravadores[ano] = pilha.enter_context(
                                pq.ParquetWriter(
                                    arquivos_anuais_temp[ano],
                                    tabela.schema,
                                    compression="zstd",
                                )
                            )
                        gravadores[ano].write_table(tabela)
            print(f"Convertido: {arquivo_zip.name}")
    arquivos_parquet_anuais = []
    for ano, caminho_temp in arquivos_anuais_temp.items():
        caminho_final = diretorio_parquet / caminho_temp.name
        caminho_temp.replace(caminho_final)
        arquivos_parquet_anuais.append(caminho_final)

print(f"Parquets anuais salvos: {len(arquivos_parquet_anuais)}")
```

```python
import re
from pathlib import Path
from tempfile import TemporaryDirectory

import pyarrow.parquet as pq

diretorio_parquet = Path("parquet_anual")
padrao_parquet_anual = re.compile(r"BTCUSDT-1s-\d{4}\.parquet$")
arquivos_parquet_anuais = sorted(
    caminho for caminho in diretorio_parquet.glob("BTCUSDT-1s-*.parquet")
    if padrao_parquet_anual.fullmatch(caminho.name)
)
if not arquivos_parquet_anuais:
    raise FileNotFoundError(
        f"Nenhum Parquet anual BTCUSDT-1s-YYYY.parquet encontrado em "
        f"{diretorio_parquet.resolve()}"
    )

arquivo_parquet_completo = diretorio_parquet / "BTCUSDT-1s-completo.parquet"
with TemporaryDirectory(dir=diretorio_parquet) as pasta_temporaria:
    caminho_temporario = Path(pasta_temporaria) / arquivo_parquet_completo.name
    schema = pq.read_schema(arquivos_parquet_anuais[0])
    with pq.ParquetWriter(caminho_temporario, schema, compression="zstd") as gravador:
        for caminho_anual in arquivos_parquet_anuais:
            leitor = pq.ParquetFile(caminho_anual)
            try:
                for bloco in leitor.iter_batches(batch_size=250_000):
                    gravador.write_batch(bloco)
            finally:
                leitor.close()

    caminho_temporario.replace(arquivo_parquet_completo)

print(f"Parquets anuais consolidados: {len(arquivos_parquet_anuais)}")
print(f"Arquivo consolidado: {arquivo_parquet_completo}")
```

```python
from pathlib import Path

import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

arquivo_parquet_completo = Path("parquet_anual") / "BTCUSDT-1s-completo.parquet"
if not arquivo_parquet_completo.is_file():
    raise FileNotFoundError(f"Arquivo Parquet n?o encontrado: {arquivo_parquet_completo.resolve()}")

# Os timestamps Spot de 2026 est?o em microssegundos; filtre no Parquet antes de carregar as linhas.
inicio_mes = pd.Timestamp("2026-08-01T00:00:00Z").value // 1_000
fim_mes = pd.Timestamp("2026-09-01T00:00:00Z").value // 1_000
colunas_grafico = ["open_time", "open", "high", "low", "close", "volume"]
klines_agosto = pd.read_parquet(
    arquivo_parquet_completo,
    columns=colunas_grafico,
    filters=[("open_time", ">=", inicio_mes), ("open_time", "<", fim_mes)],
)
if klines_agosto.empty:
    raise ValueError("O Parquet n?o cont?m candles para agosto de 2026.")

klines_agosto["datetime"] = pd.to_datetime(klines_agosto["open_time"], unit="us", utc=True)
candles_horarios = (
    klines_agosto.set_index("datetime")
    .resample("1h")
    .agg({
        "open": "first",
        "high": "max",
        "low": "min",
        "close": "last",
        "volume": "sum",
    })
    .dropna(subset=["open", "high", "low", "close"])
)

cores_volume = [
    "#2ca02c" if fechamento >= abertura else "#d62728"
    for abertura, fechamento in zip(candles_horarios["open"], candles_horarios["close"])
]
fig_candlestick = make_subplots(
    rows=2,
    cols=1,
    shared_xaxes=True,
    vertical_spacing=0.04,
    row_heights=[0.72, 0.28],
)
fig_candlestick.add_trace(
    go.Candlestick(
        x=candles_horarios.index,
        open=candles_horarios["open"],
        high=candles_horarios["high"],
        low=candles_horarios["low"],
        close=candles_horarios["close"],
        increasing_line_color="#2ca02c",
        decreasing_line_color="#d62728",
        name="BTCUSDT",
    ),
    row=1,
    col=1,
)
fig_candlestick.add_trace(
    go.Bar(
        x=candles_horarios.index,
        y=candles_horarios["volume"],
        marker_color=cores_volume,
        name="Volume (BTC)",
        hovertemplate="%{x|%d/%m %H:%M}<br>Volume: %{y:,.4f} BTC<extra></extra>",
    ),
    row=2,
    col=1,
)
fig_candlestick.update_layout(
    title="BTCUSDT ? candles de 1 hora e volume ? agosto de 2026 (UTC)",
    dragmode="pan",
    height=850,
    showlegend=False,
    margin={"l": 70, "r": 30, "t": 80, "b": 50},
)
fig_candlestick.update_yaxes(title_text="Pre?o (USDT)", row=1, col=1)
fig_candlestick.update_yaxes(title_text="Volume (BTC)", row=2, col=1)
fig_candlestick.update_xaxes(rangeslider_visible=False, row=1, col=1)
fig_candlestick.update_xaxes(
    title_text="Data e hora (UTC)",
    rangeslider_visible=True,
    rangeselector={
        "buttons": [
            {"count": 1, "label": "1d", "step": "day", "stepmode": "backward"},
            {"count": 7, "label": "7d", "step": "day", "stepmode": "backward"},
            {"step": "all", "label": "M?s inteiro"},
        ]
    },
    row=2,
    col=1,
)
fig_candlestick.show(config={"scrollZoom": True, "displaylogo": False})

print(f"Candles de 1 segundo lidos em agosto: {len(klines_agosto):,}")
print(f"Candles hor?rios no gr?fico: {len(candles_horarios):,}")
print(f"Volume total no per?odo (BTC): {candles_horarios['volume'].sum():,.4f}")
```

## Visualizar o volume dos candles

O gráfico agrega o volume dos candles por dia. A partir de 2025, os timestamps Spot da Binance são fornecidos em microssegundos; para dados anteriores, são em milissegundos.

```python
import matplotlib.pyplot as plt

klines["open_time"] = pd.to_numeric(klines["open_time"])
klines["volume"] = pd.to_numeric(klines["volume"])
unidade_tempo = "us" if klines["open_time"].max() >= 10**14 else "ms"
klines["datetime"] = pd.to_datetime(klines["open_time"], unit=unidade_tempo, utc=True)
volume_diario = klines.set_index("datetime")["volume"].resample("D").sum()

volume_diario.plot(figsize=(12, 5), title="Volume di?rio ? BTCUSDT")
plt.xlabel("Data (UTC)")
plt.ylabel("Quantidade de BTC")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
